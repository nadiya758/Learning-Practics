```python
frature=df.drop(columns="Target", axis=1)
target=df["Target"]
target
```

```python
stand = StandardScaler()
frature =  stand.fit_transform(frature)
print(frature)
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(frature,target,test_size=0.3,random_state=2)
print(X_train.shape,X_test.shape,Y_train.shape)
```

```python
model=LogisticRegression()
model.fit(X_train,Y_train)
```

```python
test_predict = model.predict(X_test)
test_predict
```

```python
accurancyTrain=accuracy_score(Y_train,model.predict(X_train))
accurancyTrain
```
