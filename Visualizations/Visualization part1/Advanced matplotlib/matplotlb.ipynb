```python
import matplotlib . pyplot as plt
import numpy as np
```

```python
x=np.linspace(0,10,100)
y=np.sin(x)
z=np.cos(x)
w=np.tan(x)
```

<h1>plotting The data

```python
plt.plot(x,y)
plt.title('sine wave')
plt.xlabel('value')
plt.ylabel('sin fo the data')
plt.show()
```

```python
plt.plot(x,z,"g D")
plt.title('cose wave')
plt.xlabel('value')
plt.ylabel('cos fo the data')
plt.show()
```

```python
plt.figure(figsize=(10,8))
plt.plot(x,y,"r--")
plt.plot(x,z, "b" )
plt.plot(x,w,"black")

plt.title(' wave')
plt.xlabel('value')
plt.ylabel(' wave fo the data')
plt.show()

```

<h1>bar plot

```python
fig=plt.figure()
ax=fig.add_axes([0,0,1,1])
lang = ['Math','English','Bangla','Phy']
peplo=[20,30,40,50]
plt.title(' Bar plot')
plt.xlabel('Language')
plt.ylabel(' peplo')
ax.pie(peplo,labels=lang,autopct='% 1.2f%%')
plt.show()


```

```python
plt.figure(figsize=(8,8))

lang = ['Math','English','Bangla','Phy']
peplo=[20,30,40,50]
plt.bar(lang,peplo)
plt.title(' Bar plot')
plt.xlabel('Language')
plt.ylabel(' peplo')
plt.show()
```
