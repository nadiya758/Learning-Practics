```python
df=pd.read_csv("/content/titanic.csv")
df
```

```python
Gender = LabelEncoder()
Embarked = LabelEncoder()
```

```python
df
```

```python
df["gender_update"]=Gender.fit_transform(df['Sex'])
df["Embarked_update"]=Embarked.fit_transform(df['Embarked'])
df
```

```python
input = df.drop(['company','job','degree','salary_more_then_100k'],axis='columns')
target =df['salary_more_then_100k']
```

```python
X_train,X_test,Y_train,Y_test = train_test_split(input,target,test_size = 0.3, random_state=2)
print(df.shape, X_train.shape, X_test.shape)
```

```python
model = DecisionTreeClassifier()
```

```python
model.fit(X_train,Y_train)
```

```python
Y_train_prediction =model.predict(X_train)
train_data_accuracy = accuracy_score(Y_train_prediction,Y_train)
print(train_data_accuracy)
```
