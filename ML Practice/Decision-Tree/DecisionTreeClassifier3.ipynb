```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
df=pd.read_csv("/content/batch_a_decision_tree_loan.csv")
df
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
labelEn = LabelEncoder()
```

```python
df["has_update"]=labelEn.fit_transform(df['has_job'])
df["loan_approved_update"]=labelEn.fit_transform(df['loan_approved'])

df
```

```python
input=df.drop(['has_job','loan_approved','loan_approved_update'],axis='columns')
target=df['loan_approved_update']
```

```python
X_train,X_test,Y_train,Y_test = train_test_split(input,target,test_size = 0.3, random_state=2)
print(df.shape, X_train.shape, X_test.shape)
```

```python
model = DecisionTreeClassifier()
```

```python
model.fit(X_train,Y_train)
```

```python
Y_train_prediction =model.predict(X_train)
train_data_accuracy = accuracy_score(Y_train_prediction,Y_train)
print(train_data_accuracy)
```

```python
Y_test_prediction =model.predict(X_test)
test_data_accuracy = accuracy_score(Y_test_prediction,Y_test)
print(test_data_accuracy)
```
