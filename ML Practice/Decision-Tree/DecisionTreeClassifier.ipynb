```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
df=pd.read_csv('salaries.csv')
df
```

```python
labelEn = LabelEncoder()
```

```python
df["company_update"]=labelEn.fit_transform(df['company'])
df["job_update"]=labelEn.fit_transform(df["job"])
df["degree_Update"]=labelEn.fit_transform(df["degree"])
df
```

```python
input=df.iloc[:,-3:]

output=df["salary_more_then_100k"]
output
```

```python
X_train,X_test,y_train,y_test=train_test_split(input,output,test_size=0.2,random_state=2)
print(df.shape,X_train.shape,X_test.shape)
```

```python
model=DecisionTreeClassifier()
model.fit(X_train,y_train)
```

```python
y_train_pred=model.predict(X_train)
Accurancy=accuracy_score(y_train,y_train_pred)
print(Accurancy)
```

```python
y_test_pred=model.predict(X_test)
Accurancy=accuracy_score(y_test,y_test_pred)
print(Accurancy)
```
