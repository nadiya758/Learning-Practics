```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
```

```python
df=pd.read_csv('diabetes.csv')
df
```

```python
target=df.iloc[:,-1]
frature=df.iloc[0:-1,:-1]

target
```

```python
scaler=StandardScaler()
frature=scaler.fit_transform()
frature
```

```python
frature=df.drop(columns="Outcome", axis=1)
target=df["Outcome"]

frature
```

```python
stand = StandardScaler()
frature =  stand.fit_transform(frature)
print(frature)
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(frature,target,test_size=0.3,random_state=2)
print(X_train.shape,X_test.shape,Y_train.shape)
```

```python
model=LogisticRegression()
model.fit(X_train,Y_train)
```

```python
test_predict = model.predict(X_test)
test_predict
```

```python
accurancyTest=accuracy_score(Y_test,test_predict)
accurancyTest
```

```python
input_data=(5,116,72,19,8,28.  8,0.587,30)
array= np.asarray(input_data)
input_data_reshape=array.reshape(1,-1)
std_data = scaler.transform(input_data_reshape)
prediction=model.predict(std_data)
if prediction==1:
  print("Diadetic")
else:
  print("Not diabetic")
```
