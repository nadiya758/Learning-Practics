```python
import numpy as np
import pandas as pd
import matplotlib .pyplot as plt
import seaborn as sns
import sklearn .datasets
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

```python
df=pd.read_csv('/content/batch_a_knn_customer_plan.csv')
df
```

```python
df.isnull().sum()
```

```python
input=df.drop(columns='plan',axis=1)
target=df['plan']
#
input
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(input,target,test_size=0.30,random_state=2)
print(input.shape,X_train.shape,X_test.shape)
```

```python
model=KNeighborsClassifier(n_neighbors=10)
model.fit(X_train,Y_train)
```

```python
Y_train_predict=model.predict(X_train)
training_data_accuracy=accuracy_score(Y_train,Y_train_predict)
training_data_accuracy
```

```python
Y_test_predict=model.predict(X_test)
test_data_accuracy=accuracy_score(Y_test,Y_test_predict)
test_data_accuracy
```
