```python
X=df.iloc[:,[3,4]]
X
```

WCSS(Within- cluster Sum 9of Squares

```python
WCSS=[]
for i in range(1,11):
  kmeans=KMeans(n_clusters=i,init='k-means++',random_state=0)
  kmeans.fit(X)
  WCSS.append(kmeans.inertia_)
WCSS
```

```python
sns. set()
plt.plot(range(1,11),WCSS)
plt.title('The Elbow Method')
plt.xlabel('Number of clusters')
plt.ylabel('WCSS')
plt.show()
```

```python
kmeans=KMeans(n_clusters=5,init='k-means++',random_state=0)
Y=kmeans.fit_predict(X)
print(Y)
```

```python
plt.figure(figsize=(10,10))
plt.scatter(X[Y==0,0],X[Y==0,1],s=50,c='red')
plt.scatter(X[Y==1,0],X[Y==1,1],s=50,c='blue')
plt.scatter(X[Y==2,0],X[Y==2,1],s=50,c='green')
plt.scatter(X[Y==3,0],X[Y==3,1],s=50,c='cyan')
plt.scatter(X[Y==4,0],X[Y==4,1],s=50,c='magenta')
plt.show()
```
