```python
class Info:
  def __init__(self,num1,num2):
    self.num1=num1
    self.num2=num2
  def jog(self):
    sum=self.num1+self.num2
    print(sum)
  def sub(self):
     sub=self.num1-self.num2
     print(sub)
  def mul(self):
      mul=self.num1*self.num2
      print(mul)
  def dev(self):
     dev=self.num1/self.num2
     print(dev)
var5=Info(10,22)
var3=Info(44,33)

print(var3.jog())
print(var5.sub())
print(var3.dev())

```

```python
class Info:
  def __init__(self,num1,num2):
    self.num1=num1
    self.num2=num2
  def jog(self):
    sum=self.num1+self.num2
    return(sum)
  def sub(self):
     sub=self.num1-self.num2
     return(sub)
  def mul(self):
      mul=self.num1*self.num2
      return(mul)
  def dev(self):
     dev=self.num1/self.num2
     return(dev)
var5=Info(10,22)
var3=Info(44,33)

print(var3.jog())
print(var5.sub())
print(var3.dev())
```

```python
class Student:
  def __init__(self,name,num1,num2,num3):
     self.name=name
     self.num1=num1
     self.num2=num2
     self.num3=num3
  def cal_average(self):
    Average=self.num1+self.num2+self.num3
    aver=Average/3
    return(aver)
var1=Student('nadia',33,44,55)
print(var1.cal_average())
```
