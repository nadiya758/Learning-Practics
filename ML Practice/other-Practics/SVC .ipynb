```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn.datasets
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
```

```python
df=sklearn.datasets.load_iris()
df
```

```python
col=df.feature_names
col
```

```python
target=df.target
target
```

```python
df_iris=pd.DataFrame(data=df.data,columns=df.feature_names)
```

```python
df_iris["target"]=df.target
df_iris
```

```python
filter_data=df_iris[df_iris.target !=2].copy()
input=filter_data.drop(["target","sepal length (cm)","sepal width (cm)"],axis=1)
output=filter_data.target
output
```

```python
sns.scatterplot(x="petal length (cm)",y="petal width (cm)",hue='target',data=df_iris)
```

```python
X_train,X_test,y_train,y_test=train_test_split(input,output,test_size=0.30,random_state=2)
print(df_iris.shape,X_test.shape,X_train.shape)
```

```python
model=SVC()
model.fit(X_train,y_train)
```

```python
y_train_predict =model.predict(X_train)
train_accuracy = accuracy_score(y_train, y_train_predict)
train_accuracy
```

```python
y_test_predict =model.predict(X_test)
test_accuracy = accuracy_score(y_test,y_test_predict)
test_accuracy
```
