### Section 4: Functions - 4 Marks

- 4.1 Check Prime Number (2 Marks)

- Scenario: You need to check whether a number is prime.

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True


num = int(input("Enter a number to check: "))
if is_prime(num):
    print("Prime")
else:
    print("Not Prime")
```

### 4.2 Calculate Average Marks (2 Marks)

- Scenario: You need to calculate average marks of a student.



```python
def calculate_average(marks):
    if not marks:
        return 0
    total = sum(marks)
    average = total / len(marks)
    return average

student_marks = [44,55,66,77,88]
print(f"Average Mark: {calculate_average(student_marks)}")
```

- 5.1 Create a Book Class (2 Marks)

- Scenario: You need to store information about a book.



```python
class Book:
    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self.price = price
my_book = Book("Poddoja", "Ilma baharoj ", 650)

# Print the book information
print(f"Title: {my_book.title}")
print(f"Author: {my_book.author}")
print(f"Price: ${my_book.price}")
```
